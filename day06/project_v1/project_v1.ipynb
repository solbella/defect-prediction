## 문제 정의

**한 문장** : 설비의 운전 조건(온도·회전수·토크·공구마모)으로, 실제 고장이 나기 전에 고장 가능성이 높은 상태를 미리 골라낸다.

| 물음 | 내 답 |
|---|---|
| 누가 쓰나 | 설비 유지보수(정비) 담당자 |
| 무엇을 결정하나 | 이 설비를 지금 멈춰서 점검·정비할지, 그대로 가동할지 |
| 언제 결정하나 | 센서값(온도·회전수·토크·공구마모)이 들어오는 가동 중, 실제 고장이 나기 전 |

# 실습 18 - 모델링 v1

## 모델링 v1

### 문법 노트 - 코드에서 찾을 이름

| 이름 | 뜻 |
|---|---|
| features | 모델에 넣는 입력 열 목록 |
| label | 정답표 - 찾으려는 쪽이면 1, 아니면 0 |
| X / y | 입력 / 정답 |
| X_train, X_test | 학습용 입력 / 시험용 입력 |
| y_train, y_test | 학습용 정답 / 시험용 정답 |
| baseline | 기준 모델 - 학습 없이 전부 아니라고 답하는 것 |
| model | 학습시킨 모델 |
| y_pred | 모델이 내놓은 답 |
| precision / recall | 정밀도 / 재현율 |

### 내 갈래

- 갈래 : [A - 맞히기(결과 열이 있다)]
- 결과 열 : [Machine failure] / 관심 갈래 : [고장]
- 이 기준으로 정한 이유 : [현장에서 걸러내려는 게 고장 셀이라서]

## Step 4. 모델링 v1

```python
# 1. 파일을 불러온다
df = pd.read_csv("../../data/16_machine-failure.csv")

# 2. 숫자로 된 열만 입력으로 쓴다. 결과 열(Machine failure)은 답이므로 뺀다
num_cols = df.select_dtypes(include="number").columns.tolist()
X = df[num_cols].copy()

# 3. 입력 열의 빈칸은 그 열의 중앙값으로 채운다
X = X.fillna(X.median())

# 4. 결과 열이 "고장"이면 1, 아니면 0인 label 열을 만든다
df["label"] = (df["Machine failure"] == "고장").astype(int)
y = df["label"]

print("입력 열:", num_cols)
print("입력 크기:", X.shape)
print("label 분포:", y.value_counts().to_dict())
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

# 5. 학습용·시험용을 8:2로 나눈다. stratify=y로 양쪽의 1 비율을 같게 맞춘다
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f"학습용 {len(X_train)}건 (고장 {y_train.sum()}건, {round(y_train.mean()*100,2)}%)")
print(f"시험용 {len(X_test)}건 (고장 {y_test.sum()}건, {round(y_test.mean()*100,2)}%)")

# 6. 기준 모델 점수 - 학습 없이 전부 0(정상)이라고만 답했을 때의 정확도
기준예측 = np.zeros(len(y_test), dtype=int)
기준정확도 = (기준예측 == y_test).mean()
print("기준 모델 정확도:", round(기준정확도 * 100, 2), "%")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score

# 7. 표준화를 한 뒤 로지스틱 회귀 모델을 학습시킨다 (학습용으로만 fit)
scaler = StandardScaler()
X_train_스케일 = scaler.fit_transform(X_train)
X_test_스케일 = scaler.transform(X_test)  # 시험용은 학습에 쓰지 않고 변환만 한다

model = LogisticRegression()
model.fit(X_train_스케일, y_train)
예측 = model.predict(X_test_스케일)

# 8. 시험용으로 일곱 줄을 출력한다
내모델정확도 = accuracy_score(y_test, 예측)
맞힌정상, 헛경보, 놓친고장, 잡은고장 = confusion_matrix(y_test, 예측).ravel()
지목한건수 = (예측 == 1).sum()
정밀도 = precision_score(y_test, 예측, zero_division=0)
재현율 = recall_score(y_test, 예측, zero_division=0)

print("기준 모델 정확도:", round(기준정확도 * 100, 2), "%")
print("내 모델 정확도:", round(내모델정확도 * 100, 2), "%")
print("지목한 건수:", 지목한건수)
print("그중 진짜:", 잡은고장)
print("시험용의 실제 고장 건수:", y_test.sum())
print("정밀도:", round(정밀도, 3))
print("재현율:", round(재현율, 3))
```

### 알아낸 것 (실행 결과 기준)

| 항목 | 값 |
|---|---|
| 기준 모델 정확도 | 96.6% |
| 내 모델 정확도 | 96.85% |
| 지목한 건수 | 13 |
| 그중 진짜 | 9 |
| 시험용의 실제 고장 건수 | 68 |
| 정밀도 | 0.692 |
| 재현율 | 0.132 |

기준 모델(전부 정상)보다 정확도는 0.25%p밖에 안 올랐고, 68건의 실제 고장 중 9건만 잡아서(재현율 0.132) 놓친 게 많다 — 앞서 "한계"에서 짚었던 대로 시험용에 고장이 68건뿐이고, 조건 열끼리 겹치는 정보가 많은 게 여기서 그대로 드러난 걸로 보인다.

```python
# 앞 모델(model)은 그대로 두고, class_weight="balanced"만 다르게 준 모델을 새로 만든다
model_가중치 = LogisticRegression(class_weight="balanced")
model_가중치.fit(X_train_스케일, y_train)
예측_가중치 = model_가중치.predict(X_test_스케일)

지목건수_가중치 = (예측_가중치 == 1).sum()
그중진짜_가중치 = ((예측_가중치 == 1) & (y_test == 1)).sum()
재현율_가중치 = recall_score(y_test, 예측_가중치, zero_division=0)

print(f"{'모델':10} {'지목 건수':>8} {'그중 진짜':>8} {'재현율':>8}")
print(f"{'손 안 댐':10} {지목한건수:>8} {잡은고장:>8} {round(재현율,3):>8}")
print(f"{'가중치':10} {지목건수_가중치:>8} {그중진짜_가중치:>8} {round(재현율_가중치,3):>8}")
```

### 결과 정리 (실행 결과 기준)
