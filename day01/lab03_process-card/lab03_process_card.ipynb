# 실습 3: 내 공정 카드 만들기

### 용어 풀이 - 반도체 8대 공정

| 순서 | 공정 | 한 줄로 |
|---|---|---|
| 1 | 웨이퍼 제조 | 실리콘으로 얇은 원판(웨이퍼)을 만들어 표면을 거울처럼 다듬는다 |
| 2 | 산화 | 웨이퍼 표면에 얇은 산화막을 입혀 보호막이자 절연막으로 삼는다 |
| 3 | 포토 (노광) | 빛으로 회로 무늬를 웨이퍼에 찍는다. 도장 찍기에 가깝다 |
| 4 | 식각 (에칭) | 찍은 무늬를 따라 필요 없는 부분을 깎아낸다 |
| 5 | 증착·이온주입 | 얇은 막을 입히고(증착), 전기 성질을 바꿀 입자를 박아 넣는다(이온주입) |
| 6 | 금속배선 | 만들어진 소자들을 금속 선으로 연결한다 |
| 7 | EDS (Electrical Die Sorting) | 자르기 전에 칩 하나하나에 전기를 흘려 합격·불합격을 판정한다 |
| 8 | 패키징 | 합격한 칩을 잘라내 포장하고 외부와 연결되게 만든다 |

2~6번은 한 번으로 끝나지 않고 층을 쌓는 만큼 여러 번 반복된다.

EDS는 재료 분야에서 쓰는 EDS(에너지 분산 분광법, Energy Dispersive Spectroscopy)와 철자만 같고 전혀 다른 말이다. 여기서는 전기를 흘려 칩을 골라내는 검사를 뜻한다.

### 식각(Etch) 공정 정리

| 항목 | 내용 |
|---|---|
| 하는 일 | 노광으로 찍은 회로 무늬를 따라, 필요 없는 막(산화막·금속·폴리실리콘 등)을 깎아낸다 |
| 측정값 | 챔버 압력(mTorr), RF 파워(W), 챔버 온도(°C), 가스 유량(sccm), 식각 시간(sec) |
| 정상 범위를 벗어나면 | 압력/파워 이상 → 과식각·저식각(패턴 손상) · 온도 이상 → 부산물 재증착, 선폭 변화 · 가스 유량 이상 → 선택비 저하, 잔류물 |
| ⚠️ 불확실 · 장비/현장별 상이 | 정상 범위 수치는 장비 제조사(LAM, Applied Materials 등)·레시피마다 다름. 위 항목은 건식(플라즈마) 식각 기준이며, 습식 식각은 온도·약액 농도·침지 시간 위주로 측정 항목 자체가 달라짐 |


## Step 3. 받은 답을 내 눈으로 검토하기

### 가스 유량 자세히 보기

| 항목 | 내용 |
|---|---|
| 단위 | sccm (분당 표준상태 cm³, standard cubic centimeter per minute) |
| 현장에서 중요하게 보는 이유 | 가스 양이 식각에 쓰이는 라디칼·이온 농도를 좌우 → 식각 속도와 선택비(원하는 막만 깎고 나머지는 보호)에 직접 영향. 부족하면 반응물이 모자라 식각이 느려지고, 과하면 잔류물이 남거나 다른 막까지 깎일 수 있음 |
| 장비/회사마다 달라지는지 | ⚠️ 확실하지 않음 — 가스 종류(CF4, Cl2, HBr 등)와 챔버 크기에 따라 기준 유량 자체가 다르고, 장비 제조사(LAM, TEL, Applied Materials 등)마다 MFC(질량유량조절기) 스펙과 레시피 기준값이 달라 정확한 수치는 실제 장비 매뉴얼·레시피 확인이 필요함 |
| **한 줄 요약** | 가스 유량은 식각 속도·선택비를 좌우하는 핵심 변수이며, 정확한 기준값은 장비·레시피마다 달라 매뉴얼 확인이 필요하다 |


## Step 4. 카드 채우기

내 공정 카드
1. 고른 공정 : 식각 (에칭)
2. 하는 일 : 찍어둔 회로 무늬를 따라 필요 없는 부분을 깎아낸다
3. 여기서 재는 값 :
   - 챔버 압력 — mTorr
   - 반응 가스 유량 — sccm
   - RF 파워 — W
   - 챔버 온도 — ℃
4. 이 값이 흔들리면 : 깎이는 정도가 웨이퍼 안에서 달라져 선폭이 들쭉날쭉해진다
5. 내 전공·경력과 닿는 지점 : 화학공학 전공 — 반응 조건(온도·압력·유량)이 결과를 좌우하는 구조가 익숙하다
6. 확인이 필요한 부분 : "식각 시간"이 측정값인지 미리 정해 넣는 설정값인지 / 챔버 온도를 어디서(척·챔버 벽 등) 재는지 — 장비마다 다를 수 있음


## Step 5. 590이라는 숫자 앞에서

내 공정 하나에서만 측정 항목이 \[5\]개.<br>
공정이 8개니까 -> [5] x 8 = [40]개.<br>
층을 쌓느라 같은 공정을 여러 번 반복한다. 여기서는 20번이라고 치고 계산해본다.<br>
-> [40] x 20 = [800]개. 590이 나오는 게 이상하지 않다.

### 챔버 압력 · 챔버 온도 자세히 보기

| 항목 | 단위 | 대략적인 범위 | 장비·회사별 차이 |
|---|---|---|---|
| 챔버 압력 | mTorr (일부 Pa 병기) | 대략 1~200 mTorr 수준 — 식각 방식(RIE/ICP 등)마다 편차가 커서 ⚠️정확한 수치는 불확실 | 있음. 식각 방식과 장비사(LAM, TEL, Applied Materials 등)·레시피마다 기준값이 다름 |
| 챔버 온도 | °C | 상온~200°C가 일반적, 일부 저온 식각은 영하대까지 냉각 — ⚠️정확한 수치는 불확실 | 있음. 상온 식각 vs 저온(크라이오) 식각, 장비마다 조절 범위가 다름 |
